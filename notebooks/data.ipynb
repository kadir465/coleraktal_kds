```python
import pandas as pd
import random
import numpy as np

# Veri setini yükle (Kendi dosya yolunu buraya yazmalısın)
dosya_yolu = r"C:\Users\kadir\OneDrive\Masaüstü\korelaktal\data\final\Sadelestirilmis_Kullanici_Verisi.csv"
df = pd.read_csv(dosya_yolu)

# --- 1. SÖZLÜKLER VE HARİTALAMA ---
# 1.0 = Erkek/Evet, 2.0 = Kadın/Hayır standart NHANES kodlaması
cinsiyet_map = {1.0: "erkek", 2.0: "kadın"}
evet_hayir_map = {1.0: "var", 2.0: "yok"}
sigara_alkol_map = {1.0: "kullanıyor", 2.0: "kullanmıyor"}

def nhanes_hasta_yorumu_olustur(row):
    # Yaş ve Cinsiyet (Eğer yaş NaN ise Bilinmeyen yazdır)
    yas = int(row['Yas']) if pd.notna(row['Yas']) else "Bilinmeyen"
    cinsiyet = cinsiyet_map.get(row['Cinsiyet'], "birey")
    
    # Alışkanlıklar
    sigara = sigara_alkol_map.get(row['Sigara_Gecmisi'], "bilinmiyor")
    alkol = sigara_alkol_map.get(row['Alkol_Tuketimi'], "bilinmiyor")
    
    # Kilo Değişimi (Eksi değerler kilo almayı, artı değerler kaybetmeyi ifade eder)
    kilo_degeri = row['Ani_Kilo_Kaybi_lbs']
    if pd.isna(kilo_degeri):
        kilo_durumu = "kilo değişimi belirtilmemiş"
    elif kilo_degeri < 0:
        kilo_durumu = f"son dönemde {abs(int(kilo_degeri))} lbs kilo almış"
    elif kilo_degeri > 0:
        kilo_durumu = f"son dönemde {int(kilo_degeri)} lbs ani kilo kaybetmiş"
    else:
        kilo_durumu = "kilosu sabit kalmış"

    # Kronik Hastalıklar
    diyabet = evet_hayir_map.get(row['Diyabet_Gecmisi'], "bilinmiyor")
    hipertansiyon = evet_hayir_map.get(row['Hipertansiyon_Gecmisi'], "bilinmiyor")
    
    # Bağırsak ve Sindirim Şikayetleri
    hareket = evet_hayir_map.get(row['Bagirsak_Hareket_Sikligi'], "belirtilmemiş")
    kacirma = evet_hayir_map.get(row['Bagirsak_Kacirma'], "belirtilmemiş")
    anormallik = evet_hayir_map.get(row['Diskida_Anormallik'], "belirtilmemiş")
    kanama = evet_hayir_map.get(row['Kanama_Sikligi'], "belirtilmemiş")

    # --- 2. ÇEŞİTLİLİK İÇİN RASTGELE METİN ŞABLONLARI ---
    sablon_secim = random.choice([1, 2, 3])
    
    if sablon_secim == 1:
        # Şablon 1: Resmi Klinik Raporu Stili
        metin = f"Klinik Not: Hasta {yas} yaşında bir {cinsiyet}. "
        metin += f"Fiziksel değişim olarak {kilo_durumu}. "
        metin += f"Alışkanlıklar değerlendirildiğinde sigara {sigara}, alkol {alkol}. "
        metin += f"Geçmiş kronik hastalık kayıtlarında diyabet {diyabet}, hipertansiyon {hipertansiyon}. "
        metin += f"Sindirim sistemi şikayetleri sorgulandığında; bağırsak hareket anormalliği {hareket}, "
        metin += f"bağırsak kaçırma {kacirma}, dışkıda şekil/renk anormalliği {anormallik} ve kanama {kanama} olarak raporlanmıştır."
        
    elif sablon_secim == 2:
        # Şablon 2: Doktor Muayene Özeti
        metin = f"Muayene Özeti: {yas} yaşındaki {cinsiyet} hastanın anamnezi alınmıştır. "
        metin += f"Hastanın kronik hastalık öyküsünde diyabet durumu {diyabet}, tansiyon problemi ise {hipertansiyon}. "
        metin += f"Madde kullanımına bakıldığında, hasta sigara {sigara} ve alkol {alkol}. Kilo profiline bakıldığında {kilo_durumu}. "
        metin += f"Lokal semptomlar şöyledir: Kanama şikayeti {kanama}, dışkıda anormallik {anormallik}, "
        metin += f"bağırsak hareket sorunu {hareket} ve kaçırma problemi {kacirma}."
        
    else:
        # Şablon 3: Hasta Beyanı Odaklı
        metin = f"Hasta Bilgisi: Birey {yas} yaşında ({cinsiyet}). "
        metin += f"Şikayetler ve geçmiş: Hasta sigara {sigara}, alkol {alkol}. "
        metin += f"Kronik olarak şeker {diyabet}, yüksek tansiyon {hipertansiyon}. "
        metin += f"Ayrıca hastanın bedensel değişiminde {kilo_durumu}. "
        metin += f"Bağırsak sağlığı detayları: Kanama {kanama}, dışkıda değişim {anormallik}, kaçırma {kacirma}, hareket anormalliği {hareket}."

    return metin.strip()

# --- 3. UYGULAMA VE KAYDETME ---
print("Metinler oluşturuluyor...")
df['Hasta_Yorumu'] = df.apply(nhanes_hasta_yorumu_olustur, axis=1)

# Label kolonunu temizle ve integer yap
df_llm = df.dropna(subset=['Hastalik_Durumu_LABEL']).copy()
df_llm['Label'] = df_llm['Hastalik_Durumu_LABEL'].astype(int)

# Sadece metin ve label kolonunu seç
df_final = df_llm[['Hasta_Yorumu', 'Label']]

# Sonucu kaydet
output_path = r"C:\Users\kadir\OneDrive\Masaüstü\korelaktal\data\final\LLM_Hazir_NHANES_Metin.csv"
df_final.to_csv(output_path, index=False)

print("\nİşlem Başarılı! Dosya kaydedildi.")
print(f"Toplam dönüştürülen satır: {len(df_final)}")
print("\n--- ÖRNEK ÇIKTILAR ---")
print("Örnek 1:", df_final['Hasta_Yorumu'].iloc[0], "| Etiket:", df_final['Label'].iloc[0])
print("Örnek 2:", df_final['Hasta_Yorumu'].iloc[1], "| Etiket:", df_final['Label'].iloc[1])
```

```python
import pandas as pd
import random

# Veriyi yüklediğini varsaydığımız dosya yolu (Kendi yolunu yazabilirsin)
dosya_yolu = r"C:\Users\kadir\OneDrive\Masaüstü\korelaktal\Sadece_Hasta_Belirtileri.csv"
df = pd.read_csv(dosya_yolu)

# 1. ETİKET (LABEL) DÖNÜŞÜMÜ: Yes -> 1 sağlıklı, No -> 0
# Boşlukları temizleyerek güvenli bir dönüşüm yapıyoruz
df['Label'] = df['Label'].astype(str).str.strip().map({'Yes': 1, 'No': 0})

# İngilizce terimleri Türkçeye çevirmek için yardımcı sözlükler
cinsiyet_sozluk = {'M': 'erkek', 'F': 'kadın'}
bmi_sozluk = {'Underweight': 'zayıf', 'Normal': 'normal kiloda', 'Overweight': 'fazla kilolu', 'Obese': 'obez'}
risk_sozluk = {'Low': 'düşük', 'Moderate': 'orta', 'High': 'yüksek'}
aktivite_sozluk = {'Low': 'hareketsiz', 'Moderate': 'orta düzeyde hareketli', 'High': 'çok hareketli'}

def hasta_yorumu_olustur(row):
    # Değerleri al ve Türkçeleştir
    yas = row.get('Age', '')
    cinsiyet = cinsiyet_sozluk.get(str(row.get('Gender')).strip(), 'birey')
    
    aile_gecmisi = True if str(row.get('Family_History')).strip() == 'Yes' else False
    sigara = True if str(row.get('Smoking_History')).strip() == 'Yes' else False
    alkol = True if str(row.get('Alcohol_Consumption')).strip() == 'Yes' else False
    diyabet = True if str(row.get('Diabetes')).strip() == 'Yes' else False
    bagirsak = True if str(row.get('Inflammatory_Bowel_Disease')).strip() == 'Yes' else False
    
    bmi = bmi_sozluk.get(str(row.get('Obesity_BMI')).strip(), 'belirsiz')
    diyet = risk_sozluk.get(str(row.get('Diet_Risk')).strip(), 'belirsiz')
    aktivite = aktivite_sozluk.get(str(row.get('Physical_Activity')).strip(), 'belirsiz')

    # Rastgele şablonlar oluşturmak için listeler
    selamlamalar = [
        "Merhaba hocam,", 
        "İyi günler doktor bey/hanım,", 
        "Hocam selamlar,", 
        "Doktor bey merhaba,", 
        "Kolay gelsin hocam,", 
        "" # Bazen direkt konuya girsin
    ]
    
    # 3 Farklı anlatım şablonu (LLM çeşitliliği için)
    sablon_tipi = random.choice([1, 2, 3])
    
    if sablon_tipi == 1:
        # Şablon 1: Hikayesel ve doğal
        metin = f"{random.choice(selamlamalar)} Ben {yas} yaşında bir {cinsiyet} hastanızım. "
        metin += f"Günlük hayatımda genel olarak {aktivite} birisiyim ve vücut kitle indeksim {bmi} seviyesinde. "
        metin += f"Beslenme alışkanlıklarımın {diyet} risk taşıdığını söyleyebilirim. "
        metin += f"Alışkanlıklarıma gelirsek; sigara {'kullanıyorum' if sigara else 'kullanmıyorum'} ve alkol {'tüketimim var' if alkol else 'almıyorum'}. "
        metin += f"Tıbbi geçmişime bakarsak, diyabetim {'var' if diyabet else 'yok'} ve inflamatuar bağırsak hastalığı şikayetim {'bulunuyor' if bagirsak else 'yok'}. "
        metin += f"Ayrıca ailemde daha önce kanser öyküsü {'var' if aile_gecmisi else 'hiç görülmedi'}."
        
    elif sablon_tipi == 2:
        # Şablon 2: Daha net ve listeleyici anlatım
        metin = f"{random.choice(selamlamalar)} Yaşım {yas}, cinsiyetim {cinsiyet}. Size sağlık durumumu özetlemek istiyorum. "
        metin += f"Ailemde kanser geçmişi {'bulunmakta' if aile_gecmisi else 'yok'}. Kendimde ise şeker hastalığı {'mevcut' if diyabet else 'yok'} "
        metin += f"ve bağırsak (inflamatuar) rahatsızlığım {'var' if bagirsak else 'bulunmuyor'}. "
        metin += f"Fiziksel olarak {aktivite} bir yaşamım var, kilom da {bmi} kategorisine giriyor. Beslenmeme dikkat etme durumum sebebiyle diyet riskim {diyet}. "
        metin += f"{'Sigara içiyorum' if sigara else 'Sigara kullanmıyorum'}, alkol durumum ise {'var' if alkol else 'yok'}."
        
    else:
        # Şablon 3: Endişeli/Soru soran hasta stili
        metin = f"{random.choice(selamlamalar)} {yas} yaşında bir {cinsiyet} olarak bazı bilgilerimi vermek istiyorum. "
        metin += f"Öncelikle ailemde kanser vakası {'görüldü' if aile_gecmisi else 'hiç olmadı'}. "
        metin += f"Ben de {'sigara içen' if sigara else 'sigara kullanmayan'} ve {'alkol alan' if alkol else 'alkol kullanmayan'} birisiyim. "
        metin += f"Kilom {bmi} düzeyinde, diyet/beslenme riskim ise {diyet} olarak ölçüldü. Hareketlilik derseniz, {aktivite} denilebilir. "
        metin += f"Kronik olarak şekerim {'var' if diyabet else 'yok'}, bunun yanında bağırsak (iltihabi) sorunum {'var' if bagirsak else 'yok'}. Durumum nedir?"

    return metin.strip()

# Fonksiyonu tüm satırlara uygula ve yeni kolonu oluştur
df['Hasta_Yorumu'] = df.apply(hasta_yorumu_olustur, axis=1)

# Sadece Hasta_Yorumu ve Label kolonlarını al
df_llm_hazir = df[['Hasta_Yorumu', 'Label']].copy()

# Eksik veya hatalı dönüşüm (NaN) olan satırları temizle
df_llm_hazir = df_llm_hazir.dropna(subset=['Label'])
df_llm_hazir['Label'] = df_llm_hazir['Label'].astype(int) # 1.0 veya 0.0 yerine tam sayı 1 ve 0 yapmak için

# Sonucu kaydet
cikis_yolu = r"C:\Users\kadir\OneDrive\Masaüstü\korelaktal\LLM_Egitim_Verisi.csv"
df_llm_hazir.to_csv(cikis_yolu, index=False)

print("LLM için metin verisi başarıyla oluşturuldu!")
print("Örnek 2 Yorum:")
print("--------------------------------------------------")
print(df_llm_hazir['Hasta_Yorumu'].iloc[0], "--> Label:", df_llm_hazir['Label'].iloc[0])
print("--------------------------------------------------")
print(df_llm_hazir['Hasta_Yorumu'].iloc[1], "--> Label:", df_llm_hazir['Label'].iloc[1])
```

```python
import pandas as pd

# Dosya yolları
path1 = r"C:\Users\kadir\OneDrive\Masaüstü\korelaktal\LLM_Egitim_Verisi.csv"
path2 = r"C:\Users\kadir\OneDrive\Masaüstü\korelaktal\data\final\LLM_Hazir_NHANES_Metin.csv"

print("Veri setleri yükleniyor...")
df1 = pd.read_csv(path1)
df2 = pd.read_csv(path2)

# 1. BİRLEŞTİRME (Concatenation)
print("Veriler birleştiriliyor...")
df_combined = pd.concat([df1, df2], ignore_index=True)

# 2. KARIŞTIRMA (Shuffling) - ÇOK ÖNEMLİ!
# frac=1 tüm veriyi al demek, random_state ise her seferinde aynı şekilde karıştırması için
print("Veriler karıştırılıyor (Shuffling)...")
df_combined = df_combined.sample(frac=1, random_state=42).reset_index(drop=True)

# 3. NİHAİ DOSYAYI KAYDETME
output_path = r"C:\Users\kadir\OneDrive\Masaüstü\korelaktal\FINAL_LLM_EGITIM_VERISI.csv"
df_combined.to_csv(output_path, index=False)

print("\n--- İŞLEM TAMAMLANDI ---")
print(f"Toplam Satır Sayısı: {len(df_combined)}")
print("\nFinal Sınıf Dağılımı:")
print(df_combined['Label'].value_counts())
```

```python
import pandas as pd

# 1. Veriyi Yükle
path = r"C:\Users\kadir\OneDrive\Masaüstü\korelaktal\FINAL_LLM_EGITIM_VERISI.csv"
df = pd.read_csv(path)

# 2. Sınıfları Ayır
df_saglikli = df[df['Label'] == 1] # Yaklaşık 147.500
df_kanserli = df[df['Label'] == 0] # Yaklaşık 68.000

# 3. Test ve Validation Setleri İçin Eşit Dağılımlı Örneklem Al (Örn: Her sınıftan 5000'er adet)
orneklem_sayisi = 12000

# Test Setini Oluştur
test_saglikli = df_saglikli.sample(n=orneklem_sayisi, random_state=42)
test_kanserli = df_kanserli.sample(n=orneklem_sayisi, random_state=42)
df_test = pd.concat([test_saglikli, test_kanserli]).sample(frac=1, random_state=42)

# Seçilenleri ana veriden çıkar (Sızıntıyı önlemek için)
df_saglikli_kalan = df_saglikli.drop(test_saglikli.index)
df_kanserli_kalan = df_kanserli.drop(test_kanserli.index)

# Validation (Doğrulama) Setini Oluştur
val_saglikli = df_saglikli_kalan.sample(n=orneklem_sayisi, random_state=42)
val_kanserli = df_kanserli_kalan.sample(n=orneklem_sayisi, random_state=42)
df_val = pd.concat([val_saglikli, val_kanserli]).sample(frac=1, random_state=42)

# Seçilenleri yine ana veriden çıkar
df_saglikli_kalan = df_saglikli_kalan.drop(val_saglikli.index)
df_kanserli_kalan = df_kanserli_kalan.drop(val_kanserli.index)

# 4. Geriye Kalan Tüm Verileri Train (Eğitim) Seti Yap
df_train = pd.concat([df_saglikli_kalan, df_kanserli_kalan]).sample(frac=1, random_state=42)

# 5. Dosyaları Kaydet
df_test.to_csv(r"C:\Users\kadir\OneDrive\Masaüstü\korelaktal\Test_Set.csv", index=False)
df_val.to_csv(r"C:\Users\kadir\OneDrive\Masaüstü\korelaktal\Validation_Set.csv", index=False)
df_train.to_csv(r"C:\Users\kadir\OneDrive\Masaüstü\korelaktal\Train_Set_Ham.csv", index=False)

# 6. İşlem Özeti
print("--- VERİ BÖLME İŞLEMİ TAMAMLANDI ---")
print(f"TEST SETİ: Toplam {len(df_test)} satır (Label 1: {len(test_saglikli)}, Label 0: {len(test_kanserli)})")
print(f"VALIDATION SETİ: Toplam {len(df_val)} satır (Label 1: {len(val_saglikli)}, Label 0: {len(val_kanserli)})")
print("\nEĞİTİM (TRAIN) SETİ HAM DURUM:")
print(f"Toplam Satır: {len(df_train)}")
print(df_train['Label'].value_counts())
```

```python
import pandas as pd

# 1. Ham Train setini yükle
df_train = pd.read_csv(r"C:\Users\kadir\OneDrive\Masaüstü\korelaktal\Train_Set_Ham.csv")

# 2. Sınıfları ayır
df_class_1 = df_train[df_train['Label'] == 1]
df_class_0 = df_train[df_train['Label'] == 0]

# 3. Azınlık sınıfı (Label 0), çoğunluk sınıfının (Label 1) sayısına eşitle
# replace=True ile eksik olan yaklaşık 80 bin satır rastgele kopyalanarak tamamlanır
df_class_0_over = df_class_0.sample(len(df_class_1), replace=True, random_state=42)

# 4. Dengeli sınıfları birleştir
df_train_balanced = pd.concat([df_class_1, df_class_0_over], axis=0)

# 5. Veriyi iyice karıştır (Kopyalar yan yana gelmemeli)
df_train_balanced = df_train_balanced.sample(frac=1, random_state=42).reset_index(drop=True)

# 6. Dengeli dosyayı kaydet
output_path = r"C:\Users\kadir\OneDrive\Masaüstü\korelaktal\Train_Set_Dengeli.csv"
df_train_balanced.to_csv(output_path, index=False)

# 7. Sonuçları yazdır
print("--- TRAIN SETİ DENGELEME İŞLEMİ TAMAMLANDI ---")
print(f"Yeni Toplam Satır Sayısı: {len(df_train_balanced)}")
print("\nYeni Sınıf Dağılımı:")
print(df_train_balanced['Label'].value_counts())
```

```python
import pandas as pd
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import AutoTokenizer, AutoModelForSequenceClassification, get_linear_schedule_with_warmup
from torch.optim import AdamW
from sklearn.metrics import accuracy_score, classification_report

# 1. Cihaz Ayarı (GPU varsa harika olur, yoksa CPU)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Kullanılan Cihaz: {device}")

# 2. Tokenizer ve Model Seçimi (Türkçe destekli BERT)
model_name = "dbmdz/bert-base-turkish-cased"
tokenizer = AutoTokenizer.from_pretrained(model_name)

# 3. Özel Dataset Sınıfı
class KanserDataset(Dataset):
    def __init__(self, texts, labels, tokenizer, max_len=128):
        self.texts = texts
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_len = max_len

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, item):
        text = str(self.texts[item])
        label = self.labels[item]

        encoding = self.tokenizer.encode_plus(
            text,
            add_special_tokens=True,
            max_length=self.max_len,
            padding='max_length',
            truncation=True,
            return_attention_mask=True,
            return_tensors='pt',
        )

        return {
            'input_ids': encoding['input_ids'].flatten(),
            'attention_mask': encoding['attention_mask'].flatten(),
            'labels': torch.tensor(label, dtype=torch.long)
        }

# 4. Verileri Yükle ve Hazırla
train_df = pd.read_csv(r"C:\Users\kadir\OneDrive\Masaüstü\korelaktal\Train_Set_Balanced.csv")
val_df = pd.read_csv(r"C:\Users\kadir\OneDrive\Masaüstü\korelaktal\Validation_Set.csv")

# Küçük bir örneklemle (deneme için) veya tamamıyla başla
train_dataset = KanserDataset(train_df['Hasta_Yorumu'].values, train_df['Label'].values, tokenizer)
val_dataset = KanserDataset(val_df['Hasta_Yorumu'].values, val_df['Label'].values, tokenizer)

train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=16)

# 5. Modeli Tanımla (2 Sınıf: Sağlıklı / Kanserli)
model = AutoModelForSequenceClassification.from_pretrained(model_name, num_labels=2)
model.to(device)

print("Model ve Tokenizasyon hazır!")
```

```python
# Basit bir Eğitim Döngüsü (Training Loop)
optimizer = AdamW(model.parameters(), lr=2e-5) # Öğrenme hızı

model.train()
for epoch in range(3): # Verinin üzerinden 3 kez geç
    total_loss = 0
    for batch in train_loader:
        optimizer.zero_grad()
        
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        labels = batch['labels'].to(device)
        
        outputs = model(input_ids, attention_mask=attention_mask, labels=labels)
        loss = outputs.loss
        total_loss += loss.item()
        
        loss.backward()
        optimizer.step()
        
    print(f"Epoch {epoch+1} tamamlandı. Ortalama Kayıp (Loss): {total_loss/len(train_loader)}")
```
