```python
import pandas as pd

# Dosya yollarını kendi sistemine göre düzenle
file_path = r'C:\Users\kadir\OneDrive\Masaüstü\korelaktal\colorectal_cancer_dataset.csv'
output_path = r'C:\Users\kadir\OneDrive\Masaüstü\korelaktal\data\raw\2005-2006\Kullanici_Belirtileri_Risk_Seti.csv'

print("Veri seti yükleniyor...")
df = pd.read_csv(file_path, low_memory=False)

# 1. ADIM: Özellikleri (Features) ve Etiketi (Label) Belirleme
# Sadece kullanıcının beyan edebileceği risk faktörlerini seçiyoruz
ozellikler_listesi = [
    'Age',
    'Gender',
    'Family_History',
    'Smoking_History',
    'Alcohol_Consumption',
    'Obesity_BMI',
    'Diet_Risk',
    'Physical_Activity',
    'Diabetes',
    'Inflammatory_Bowel_Disease',
    'Genetic_Mutation',
    'Screening_History'
]

hedef_etiket = 'Survival_Prediction'

# Veri setinde mevcut olan kolonları kontrol ederek seçme işlemi
mevcut_ozellikler = [col for col in ozellikler_listesi if col in df.columns]
secilecek_kolonlar = ['Patient_ID'] + mevcut_ozellikler + [hedef_etiket]

# Sadece bu kolonları içeren yeni dataframe oluşturma
df_final = df[[col for col in secilecek_kolonlar if col in df.columns]].copy()

# 2. ADIM: Temizlik
# Tahmin hedefi (Label) boş olan satırları model eğitemeyeceğimiz için siliyoruz
df_final = df_final.dropna(subset=[hedef_etiket])

# 3. ADIM: Kaydetme
df_final.to_csv(output_path, index=False)

print("\n--- İŞLEM ÖZETİ ---")
print(f"Toplam Gözlem (Satır) Sayısı: {len(df_final)}")
print(f"Girdi Özellikleri (Features) Sayısı: {len(mevcut_ozellikler)}")
print(f"Hedef Değişken (Label): {hedef_etiket}")

print("\nSeçilen Özellikler:")
print(mevcut_ozellikler)

if hedef_etiket in df_final.columns:
    print("\nEtiket (Label) Dağılımı:")
    print(df_final[hedef_etiket].value_counts())
```
