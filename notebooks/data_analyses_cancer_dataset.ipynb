```python
import pandas as pd

# 1. Veri setini okuyalım
df = pd.read_csv(r"C:\Users\kadir\OneDrive\Masaüstü\korelaktal\colorectal_cancer_dataset.csv")

# 2. Kanser evrelerini sayısallaştırmak için Label sütunu oluşturalım
# Başlangıç (Localized) = 0, Orta (Regional) = 1, İleri (Metastatic) = 2
evre_sozlugu = {
    "Localized": 0,
    "Regional": 1,
    "Metastatic": 1
}
df['Label'] = df['Cancer_Stage'].map(evre_sozlugu)

# 3. Modele verilmemesi gereken (Geleceğe dair veya gereksiz) sütunların listesi
silinecek_sutunlar = [
    'Patient_ID', 
    'Cancer_Stage',             # Artık Label yaptık, orijinalini siliyoruz
    'Tumor_Size_mm',            # Erken teşhiste tümör boyutu bilinmez (Veri sızıntısı yapar)
    'Survival_5_years',         # Geleceğe dair sonuç (Tahmin edilmemeli)
    'Mortality',                # Geleceğe dair sonuç
    'Country',                  # Hastalıkla doğrudan ilgisi olmayan makro veriler
    'Healthcare_Costs', 
    'Incidence_Rate_per_100K', 
    'Mortality_Rate_per_100K', 
    'Urban_or_Rural', 
    'Economic_Classification', 
    'Healthcare_Access', 
    'Insurance_Status'
]

# 4. Bu sütunları tablodan tamamen çıkaralım
# errors='ignore' parametresi, eğer tabloda o sütun yoksa hata vermeden geçmesini sağlar
df_egitim = df.drop(columns=silinecek_sutunlar, errors='ignore')

# 5. Eğitime hazır verinin ilk 5 satırını ve kalan sütunları görelim
print(df_egitim.head())
print("\nKalan Sütunlar (Özellikler + Label):")
print(df_egitim.columns.tolist())
```

```python
import pandas as pd
import random

def dogal_metin_olustur(row):
    metin_parcalari = []
    
    # 1. Çeşitli Giriş ve Demografi Kalıpları
    giris_kaliplari = [
        f"Merhaba, ben {row['Age']} yaşında bir {'erkek' if row['Gender'] == 'M' else 'kadın'} hastayım.",
        f"İyi günler. Yaşım {row['Age']}, cinsiyetim {'erkek' if row['Gender'] == 'M' else 'kadın'}.",
        f"{row['Age']} yaşındayım ({'Erkek' if row['Gender'] == 'M' else 'Kadın'}).",
        f"Doktor bey/hanım merhabalar, {row['Age']} yaşındayım.",
        f"Kontrol amaçlı yazıyorum. {row['Age']} yaşında bir {'erkeğim' if row['Gender'] == 'M' else 'kadınım'}."
    ]
    metin_parcalari.append(random.choice(giris_kaliplari))

    # 2. Aile Öyküsü
    if row['Family_History'] == 'Yes':
        aile_kaliplari = ["Ailemde kanser geçmişi var.", "Genetik olarak ailede kanser öykümüz bulunuyor.", "Daha önce aile üyelerimde kanser görüldü."]
        metin_parcalari.append(random.choice(aile_kaliplari))
    else:
        aile_kaliplari = ["Ailemde bilinen bir kanser vakası yok.", "Birinci derece akrabalarımda kanser öyküsü bulunmuyor.", "Genetik bir kanser geçmişimiz yok."]
        metin_parcalari.append(random.choice(aile_kaliplari))

    # 3. Kötü Alışkanlıklar (Sigara & Alkol)
    aliskanlik = ""
    if row['Smoking_History'] == 'Yes' and row['Alcohol_Consumption'] == 'Yes':
        aliskanlik = random.choice(["Hem sigara hem de alkol kullanıyorum.", "Sigara ve alkol tüketimim var."])
    elif row['Smoking_History'] == 'Yes':
        aliskanlik = random.choice(["Sigara içiyorum ama alkol almıyorum.", "Sadece sigara kullanıyorum."])
    elif row['Alcohol_Consumption'] == 'Yes':
        aliskanlik = random.choice(["Sigara içmiyorum fakat alkol tüketiyorum.", "Sadece alkol kullanımım var."])
    else:
        aliskanlik = random.choice(["Sigara veya alkol gibi kötü alışkanlıklarım yok.", "Hayatım boyunca sigara ve alkolden uzak durdum."])
    metin_parcalari.append(aliskanlik)

    # 4. Kilo ve Diyet Durumu
    if row['Obesity_BMI'] == 'Obese':
        kilo = random.choice(["Ciddi kilo problemim var.", "Obezite sınırındayım.", "Kilom normalin oldukça üzerinde."])
    elif row['Obesity_BMI'] == 'Overweight':
        kilo = random.choice(["Biraz fazla kilom var.", "Kilo vermeye çalışıyorum.", "Hafif kilolu bir yapım var."])
    else:
        kilo = random.choice(["Kilom gayet normal seviyelerde.", "Kilo problemim bulunmuyor.", "İdeal kilomdayım."])
    
    if row['Diet_Risk'] == 'High':
        kilo += " " + random.choice(["Beslenmeme hiç dikkat etmiyorum.", "Genellikle sağlıksız ve yağlı beslenirim."])
    
    metin_parcalari.append(kilo)

    # 5. Kronik Hastalıklar (Diyabet ve İltihabi Bağırsak)
    kronik = []
    if row['Diabetes'] == 'Yes':
        kronik.append("şeker hastalığım")
    if row['Inflammatory_Bowel_Disease'] == 'Yes':
        kronik.append("iltihabi bağırsak rahatsızlığım")
        
    if kronik:
        hastalik_metni = random.choice([
            f"Ayrıca bende {' ve '.join(kronik)} var.", 
            f"Bilinen kronik olarak {' ve '.join(kronik)} ile mücadele ediyorum."
        ])
        metin_parcalari.append(hastalik_metni)

    # 6. Tarama Geçmişi (Screening_History)
    if row['Screening_History'] == 'Regular':
        tarama = random.choice(["Düzenli olarak sağlık taramalarımı ve kolonoskopimi yaptırırım.", "Doktor kontrollerimi hiç aksatmam."])
    elif row['Screening_History'] == 'Irregular':
        tarama = random.choice(["Taramalarımı düzenli yaptırdığım söylenemez.", "Ara sıra doktora giderim."])
    else:
        tarama = random.choice(["Daha önce hiç kanser taraması yaptırmadım.", "Kolonoskopi veya benzeri bir testten geçmedim."])
    metin_parcalari.append(tarama)

    # Parçaları boşlukla birleştirerek doğal bir paragraf oluştur
    return " ".join(metin_parcalari)

# Veri çerçevenize (df) yeni sütunu uygulama
df['Hasta_Yorumu'] = df.apply(dogal_metin_olustur, axis=1)

# Örnek ilk 5 satırın sonucunu görmek için:
# for index, row in df.head().iterrows():
#     print(f"Satır {index}:", dogal_metin_olustur(row))
#     print("-" * 50)
```
